```python
# =============================================================================
# Practical 4: Real-time Object Detection with YOLOv8 (YouTube Shorts)
# =============================================================================

from ultralytics import YOLO
import cv2
import os
import yt_dlp
from IPython.display import HTML
from base64 import b64encode

print("Ultralytics YOLOv8 ready")

# ========================================================================
# Step 0: Configure FFmpeg Path (ADD THIS!)
# ========================================================================

# Add FFmpeg to system PATH
FFMPEG_PATH = r"C:\ffmpeg\bin"
os.environ["PATH"] = FFMPEG_PATH + os.pathsep + os.environ["PATH"]

# Verify FFmpeg is accessible
import subprocess
try:
    result = subprocess.run(
        ["ffmpeg", "-version"], 
        capture_output=True, 
        text=True, 
        shell=True
    )
    if result.returncode == 0:
        print("✓ FFmpeg found and working")
    else:
        print("⚠️ FFmpeg found but may have issues")
except Exception as e:
    print(f"❌ FFmpeg not found: {e}")
    print(f"   Make sure FFmpeg is installed at: {FFMPEG_PATH}")

# ========================================================================
# Step 1: Define paths
# ========================================================================

DATASET_PATH = r"C:\Users\Nakul\Downloads\CV\cnn\cnn\Images"
MODEL_PATH = os.path.join("runs", "detect", "your_custom_model", "weights", "best.pt")

OUTPUT_DIR = r"C:\Users\Nakul\Downloads\CV\cnn\cnn"
os.makedirs(OUTPUT_DIR, exist_ok=True)  # Create directory if doesn't exist

INPUT_VIDEO = os.path.join(OUTPUT_DIR, "shorts_video.mp4")
OUTPUT_VIDEO = os.path.join(OUTPUT_DIR, "annotated_shorts.mp4")

TRAIN_NAME = "your_custom_model"

# ========================================================================
# Step 2: Train custom YOLOv8 model (optional)
# ========================================================================

# Uncomment to train custom model
"""
print("Starting training...")
model = YOLO("yolov8n.pt")
model.train(
    data=os.path.join(DATASET_PATH, "data.yaml"),
    epochs=50,
    imgsz=640,
    batch=16,
    name=TRAIN_NAME
)
print("Training finished!")
"""

# ========================================================================
# Step 3: Load model
# ========================================================================

if os.path.exists(MODEL_PATH):
    print(f"Loading custom model from: {MODEL_PATH}")
    model = YOLO(MODEL_PATH)
else:
    print("Custom model not found, loading pretrained yolov8n.pt")
    model = YOLO("yolov8n.pt")

print("✓ Model loaded")

# ========================================================================
# Step 4: Download YouTube Shorts (WITH FFMPEG CONFIG)
# ========================================================================

YOUTUBE_URL = "https://www.youtube.com/shorts/G5vp6kgZy1E"

# Updated yt-dlp options with FFmpeg path
ydl_opts = {
    "format": "best[ext=mp4]/best",  # Simplified format
    "outtmpl": INPUT_VIDEO,
    "merge_output_format": "mp4",
    "ffmpeg_location": FFMPEG_PATH,  # ← ADDED: Point to FFmpeg
    "quiet": False,
    "no_warnings": False,
}

print(f"Downloading YouTube Shorts: {YOUTUBE_URL}")

try:
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([YOUTUBE_URL])
    print(f"✓ Downloaded video: {INPUT_VIDEO}")
except Exception as e:
    print(f"❌ Download failed: {e}")
    print("\n💡 Troubleshooting:")
    print(f"   1. Check FFmpeg is at: {FFMPEG_PATH}")
    print("   2. Try running: ffmpeg -version in command prompt")
    print("   3. Restart your Python kernel")
    exit()

# ========================================================================
# Step 5: Run YOLOv8 and save annotated video
# ========================================================================

cap = cv2.VideoCapture(INPUT_VIDEO)

if not cap.isOpened():
    print(f"❌ Error: Cannot open video file: {INPUT_VIDEO}")
    exit()

# Get video properties
fps = int(cap.get(cv2.CAP_PROP_FPS))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print(f"\n📊 Video Info:")
print(f"   Resolution: {width}x{height}")
print(f"   FPS: {fps}")
print(f"   Total frames: {total_frames}")
print(f"   Duration: {total_frames/fps:.1f} seconds")

# Create VideoWriter
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(OUTPUT_VIDEO, fourcc, fps, (width, height))

print("\n🎬 Running YOLOv8 detection...")

frame_count = 0
while True:
    ret, frame = cap.read()
    if not ret:
        break

    # YOLOv8 inference
    results = model(frame, verbose=False, conf=0.25)  # 25% confidence threshold
    
    # Draw bounding boxes
    annotated_frame = results[0].plot()
    
    # Write to output
    out.write(annotated_frame)
    
    frame_count += 1
    if frame_count % 30 == 0:  # Progress update
        progress = (frame_count / total_frames) * 100
        print(f"   Processed: {frame_count}/{total_frames} frames ({progress:.1f}%)")

cap.release()
out.release()

print(f"\n✅ Annotated video saved: {OUTPUT_VIDEO}")

# ========================================================================
# Step 6: Display annotated video
# ========================================================================

print("\n📺 Displaying annotated video:")

try:
    # Convert video to base64 for HTML display
    with open(OUTPUT_VIDEO, "rb") as f:
        video_bytes = f.read()
    video_b64 = b64encode(video_bytes).decode()

    video_html = f"""
    <video width="800" controls>
      <source src="data:video/mp4;base64,{video_b64}" type="video/mp4">
      Your browser does not support the video tag.
    </video>
    """
    
    display(HTML(video_html))
    
except Exception as e:
    print(f"⚠️ Could not display video in browser: {e}")
    print(f"📂 Open manually: {OUTPUT_VIDEO}")

# ========================================================================
# Step 7: Statistics
# ========================================================================

print("\n📊 Detection Statistics:")
print(f"   Total frames processed: {frame_count}")
print(f"   Output video: {OUTPUT_VIDEO}")
print(f"   File size: {os.path.getsize(OUTPUT_VIDEO)/1024/1024:.2f} MB")

print("\n✅ COMPLETE!")

# ========================================================================
# Optional: Run on single image
# ========================================================================

# Example: Detect objects in a single image
"""
results = model("path/to/your_image.jpg")
results[0].save("result.jpg")
print("Saved result to result.jpg")
"""
```
